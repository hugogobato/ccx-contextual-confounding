```
import os, sys, time, json, hashlib
import numpy as np
import pandas as pd
CODE_HASH = hashlib.sha256(('import os, sys, time, json, hashlib\nimport numpy as np\nimport pandas as pd\n\n"""WP 3.1 continuous-outcome witnesses (Phase 3 flagship, claim C4).\n\nConstructions (predeclared design; see memos/gate_C_memo.md):\n\nContext cover: X-strata (quantile bins of the treatment/residual covariate),\nK strata by n (K = 3..8). Within each context c, ADDITIVE-MECHANISM\nresiduals r = y - m_hat(x) form an empirical law mu_c.\n\nDeclared structural null S_cont (homoskedastic shared-shape ANM):\ny = m(x) + sigma * eps, eps ~ F shared, eps independent of x. After\nper-context affine standardization (center/scale by context residual\nmoments), every context law must COINCIDE with one common shape. Latent\ncommon causes break this: U enters both arms, so stratum-restricted U\nranges inject stratum-dependent variance/skew into standardized residuals.\n\nK1 (kernelized forcing-divergence witness):\n  T_K1 = min_{w in simplex} sum_c MMD_k^2( mu_hat_c , nu_c(w) ),\nwhere nu_c(w) is the Gaussian-smoothed mixture with weights w over a fixed\ncomponent grid t_1..t_G (standard-normal quantiles) pushed through the\ncontext affine map. Mean embeddings are linear in w => exact QP per sum;\nsolved by projected gradient (Frank-Wolfe) to machine tolerance. Value 0\niff ONE shared coupling explains all context laws simultaneously (in the\nkernel-smoothed sense) - the continuous descendant of the discrete KL\ncontextuality functional (KL -> MMD^2 forcing divergence).\n\nK2 (transport-defect witness):\n  Consecutive-context squared Wasserstein distances d_c =\n  W2^2(mu_c, mu_{c+1}) (exact 1-D empirical OT). Under the shared-shape\n  null with drift, d_c ~ (m_{c+1} - m_c)^2 up to sampling error; the\n  witness is the max standardized EXCESS transport\n  T_K2 = max_c (d_c - fitted)/se, the cycle-space/monodromy-excess\n  descendant of B2\'s topological statistic, requiring no support holes.\n\nCalibration: multiplier (Rademacher wild) bootstrap recomputing the whole\npipeline on multiplier-perturbed residuals, including the inner\noptimization; optional trimming policy applied BEFORE context moments\n(predeclared grids q in {0, 0.01, 0.05}).\n\nAll estimators take arrays (x, y) and return scalar statistics; bootstrap\nvariants return draws for critical values.\n"""\nimport numpy as np\n\ntry:\n    import ot as pot\nexcept Exception:                                   # pragma: no cover\n    pot = None\n\n\n# ------------------------------------------------------------------ contexts\n\ndef make_contexts(x, K=None, return_edges=False):\n    """Quantile-stratify x into K contexts. Default K by n: 3 (<800),\n    5 (<4000), 8 otherwise (predeclared)."""\n    x = np.asarray(x, dtype=float)\n    n = len(x)\n    if K is None:\n        K = 3 if n < 800 else (5 if n < 4000 else 8)\n    qs = np.linspace(0, 1, K + 1)[1:-1]\n    edges = np.quantile(x, qs)\n    ctx = np.searchsorted(edges, x, side="right")\n    if return_edges:\n        return ctx, edges, K\n    return ctx\n\n\ndef context_residuals(x, y, ctx, K, trim_q=0.0):\n    """Additive-mechanism residuals per context: r = y - context-mean of y\n    (context-constant m_hat; documented estimator). Winsorizes residuals at\n    trim_q/1-trim_q WITHIN context when trim_q > 0."""\n    out = []\n    for c in range(K):\n        m = ctx == c\n        r = y[m] - y[m].mean()\n        if trim_q > 0 and len(r) >= 10:\n            lo, hi = np.quantile(r, [trim_q, 1 - trim_q])\n            r = np.clip(r, lo, hi)\n        out.append(r)\n    return out\n\n\ndef _median_bandwidth(resids):\n    pool = np.concatenate(resids)\n    pool = pool[np.isfinite(pool)]\n    if len(pool) < 10:\n        return 1.0\n    sub = pool[::max(1, len(pool) // 2000)]\n    d = np.abs(sub[:, None] - sub[None, :])\n    med = np.median(d[d > 0]) if np.any(d > 0) else 1.0\n    return max(med * 1.06, 1e-6)\n\n\ndef _gram(points, bw):\n    """RBF Gram matrix and cross matrices against component grid."""\n    z = (points[:, None] - points[None, :]) / bw\n    return np.exp(-0.5 * z ** 2)\n\n\n# ------------------------------------------------------------------ K1 core\n\n_GRID_CACHE = {}\n\n\ndef _component_grid(G):\n    """Standard-normal quantile grid (deterministic rng(7)); cached per G\n    since it is identical across every construction."""\n    if G not in _GRID_CACHE:\n        _GRID_CACHE[G] = np.quantile(\n            np.random.default_rng(7).normal(size=200000),\n            (np.arange(G) + 0.5) / G)\n    return _GRID_CACHE[G]\n\n\nclass K1Witness:\n    """Kernelized forcing-divergence witness over a shared coupling grid."""\n\n    def __init__(self, resids, G=32, bw=None, fw_iters=300, m_cap=600):\n        self.K = len(resids)\n        self.G = G\n        self.fw_iters = fw_iters\n        self.bw = bw if bw is not None else _median_bandwidth(resids)\n        self.m_cap = m_cap\n        # component grid: standard normal quantiles\n        self.t = _component_grid(G)\n        # per-context mean embeddings against grid components and self-Gram\n        self.b = []      # b_c[j] = <mu_c, nu_j>\n        self.A = []      # A_c[j,j\'] = <nu_j, nu_j\'> (same for all c)\n        self.self_mmd_const = []\n        Ky = _gram(self.t, self.bw)\n        self.A = Ky.copy()\n        for r in resids:\n            if len(r) == 0:\n                r = np.zeros(1)\n            rs = r[:min(len(r), self.m_cap)]\n            k_cross = np.exp(-0.5 * ((rs[:, None] - self.t[None, :]) /\n                                     self.bw) ** 2).mean(axis=0)\n            self.b.append(k_cross)\n            Krr = _gram(rs[:min(len(rs), 400)], self.bw)\n            self.self_mmd_const.append(float(Krr.mean()))\n\n    def objective(self, w, c=None):\n        """sum_c MMD^2(mu_c, nu_c(w)) (constants included)."""\n        idx = range(self.K) if c is None else [c]\n        tot = 0.0\n        for cc in idx:\n            tot += float(w @ self.A @ w - 2 * self.b[cc] @ w +\n                         self.self_mmd_const[cc])\n        return tot\n\n    def solve(self, w_init=None, smo_rounds=None):\n        """Frank-Wolfe on the simplex for min_w sum_c MMD^2(mu_c, nu_c(w)).\n        Objective (up to constants): f(w) = K * w\'Aw - 2 * (sum_c b_c)\'w.\n        Exact line-search FW followed by pairwise (SMO) refinement."""\n        s = np.sum(self.b, axis=0)\n        w = np.ones(self.G) / self.G if w_init is None else \\\n            np.asarray(w_init, float).copy()\n        w = np.maximum(w, 0)\n        w /= max(w.sum(), 1e-300)\n        prev = np.inf\n        for k in range(self.fw_iters):\n            grad = 2 * self.K * (self.A @ w) - 2 * s\n            j = int(np.argmin(grad))\n            d = np.zeros(self.G)\n            d[j] = 1.0\n            # exact step via line search on the segment (convex quadratic):\n            num = float(grad @ (d - w))\n            den = float(2 * self.K * (d - w) @ self.A @ (d - w))\n            gamma = 0.0 if den <= 0 else max(0.0, min(1.0,\n                                                      -num / den))\n            if gamma == 0.0:\n                break\n            w = (1 - gamma) * w + gamma * d\n            val = self.objective(w)\n            if abs(prev - val) < 1e-14 * max(1.0, abs(val)):\n                break\n            prev = val\n        w = self._pairwise_polish(w, max_rounds=smo_rounds or 300)\n        self.w_star = np.maximum(w, 0)\n        self.w_star /= max(self.w_star.sum(), 1e-300)\n        self.value = max(self.objective(self.w_star), 0.0)\n        return self.value\n\n    def _pairwise_polish(self, w, max_rounds=300, tol=1e-15):\n        """Exact pairwise coordinate optimization over the simplex:\n        repeatedly move mass along the best pair direction e_i - e_j with\n        closed-form step, guaranteeing monotone descent to the QP optimum."""\n        s = np.sum(self.b, axis=0)\n        K = self.K\n        prev = self.objective(w)\n        for _ in range(max_rounds):\n            g = 2 * K * (self.A @ w - s / K)\n            # d_ij = e_i - e_j ; decrease along gamma:\n            # f(w + g*d) = f + gamma*(g_i - g_j) + gamma^2 * K * c_ij\n            Gd = self.A  # quadratic coefficients via diag/offdiag\n            gi = g[:, None] - g[None, :]\n            cij = np.diag(Gd)[:, None] + np.diag(Gd)[None, :] \\\n                - 2 * Gd\n            den = 2 * K * np.maximum(cij, 0.0)\n            with np.errstate(divide="ignore", invalid="ignore"):\n                gam = -gi / np.where(den > 0, den, 1.0)\n            hi = np.tile(w, (self.G, 1))          # gamma <= w_j\n            lo = -np.tile(w[:, None], (1, self.G))  # gamma >= -w_i\n            gam = np.clip(np.where(den > 0, gam, 0.0),\n                          np.minimum(lo, 0.0) - 1e-12,\n                          np.maximum(hi, 0.0) + 1e-12)\n            dec = gi * gam + 0.5 * den * gam ** 2\n            np.fill_diagonal(dec, 0.0)\n            i, j = np.unravel_index(np.argmin(dec), dec.shape)\n            best_dec = -dec[i, j]\n            if best_dec <= tol * max(1.0, abs(prev)):\n                break\n            gamma = gam[i, j]\n            w = w.copy()\n            w[i] += gamma\n            w[j] -= gamma\n            prev -= best_dec\n        return w\n\n\ndef k1_witness(x, y, trim_q=0.0, G=32, K=None, return_model=False):\n    """Point statistic T_K1."""\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    resids = context_residuals(x, y, ctx, Kr, trim_q=trim_q)\n    model = K1Witness(resids, G=G)\n    val = model.solve()\n    if return_model:\n        return val, model\n    return val\n\n\ndef k1_multiplier_bootstrap(x, y, B=199, trim_grid=(0.0,), G_boot=16, K=None,\n                            rng=None, bmap=None):\n    """Wild-bootstrap draws of T_K1 (Rademacher multipliers on residuals,\n    pipeline re-run incl. inner optimization). bmap: optional\n    {trim_q: B_eff} per-trim draw budgets (D8). Returns (B,) array for the\n    FIRST trimming level plus dict of arrays per trimming level."""\n    rng = rng or np.random.default_rng(20260827)\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    base_resids = context_residuals(x, y, ctx, Kr, trim_q=0.0)\n    base_model = K1Witness(base_resids, G=G_boot)\n    base_model.solve()\n    w_star = base_model.w_star\n    draws = {}\n    for tq in trim_grid:\n        b_eff = B if bmap is None else int(bmap.get(tq, B))\n        vals = np.empty(b_eff)\n        for b in range(b_eff):\n            pert = [(r * rng.choice([-1.0, 1.0], size=len(r))) for r in\n                    base_resids]\n            if tq > 0:\n                pert = [_winsor(r, tq) for r in pert]\n            m = K1Witness(pert, G=G_boot, bw=base_model.bw)\n            vals[b] = m.solve(w_init=w_star, smo_rounds=80)\n        draws[tq] = vals\n    return draws[list(trim_grid)[0]], draws\n\n\ndef _winsor(r, q):\n    lo, hi = np.quantile(r, [q, 1 - q])\n    return np.clip(r, lo, hi)\n\n\n# ------------------------------------------------------------------ K2 core\n\ndef _w2sq(a, b):\n    """Exact 1-D empirical squared Wasserstein-2 (sort-based)."""\n    a = np.sort(np.asarray(a, float))\n    b = np.sort(np.asarray(b, float))\n    n = min(len(a), len(b))\n    if n < 2:\n        return 0.0\n    ai = np.quantile(a, np.linspace(0, 1, n))\n    bi = np.quantile(b, np.linspace(0, 1, n))\n    return float(((ai - bi) ** 2).mean())\n\n\ndef k2_witness(x, y, trim_q=0.0, K=None, return_parts=False):\n    """Transport-defect witness: max standardized excess of consecutive-\n    context W2^2 over the location-gap prediction."""\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    resids = context_residuals(x, y, ctx, Kr, trim_q=trim_q)\n    means = np.array([r.mean() for r in resids])\n    vars_ = np.array([r.var() for r in resids])\n    d = np.array([_w2sq(resids[c], resids[c + 1]) for c in range(Kr - 1)])\n    gap = np.diff(means) ** 2\n    if len(d) < 2:\n        val = 0.0\n    else:\n        # excess over drift prediction, standardized by pooled sampling scale\n        resid_excess = np.maximum(d - gap, 0.0)\n        se = np.sqrt((vars_[:-1]**2 / np.array([len(r) for r in resids[:-1]])\n                      + vars_[1:]**2 /\n                      np.array([len(r) for r in resids[1:]])) / 2.0) * 2.0\n        se = np.maximum(se, 1e-12)\n        val = float(np.max(resid_excess / se))\n    if return_parts:\n        return val, {"d": d, "gap": gap, "means": means, "vars": vars_}\n    return val\n\n\ndef k2_multiplier_bootstrap(x, y, B=199, trim_grid=(0.0,), K=None,\n                            rng=None, bmap=None):\n    """bmap: optional {trim_q: B_eff} per-trim draw budgets (D8)."""\n    rng = rng or np.random.default_rng(20260828)\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    base = context_residuals(x, y, ctx, Kr, trim_q=0.0)\n    draws = {}\n    for tq in trim_grid:\n        b_eff = B if bmap is None else int(bmap.get(tq, B))\n        vals = np.empty(b_eff)\n        for b in range(b_eff):\n            pert = [(r * rng.choice([-1.0, 1.0], size=len(r))) for r in base]\n            if tq > 0:\n                pert = [_winsor(r, tq) for r in pert]\n            vals[b] = k2_from_resids(pert)\n        draws[tq] = vals\n    return draws[list(trim_grid)[0]], draws\n\n\ndef k2_from_resids(resids):\n    means = np.array([r.mean() for r in resids])\n    vars_ = np.array([r.var() for r in resids])\n    d = np.array([_w2sq(resids[c], resids[c + 1])\n                  for c in range(len(resids) - 1)])\n    gap = np.diff(means) ** 2\n    if len(d) < 2:\n        return 0.0\n    ex = np.maximum(d - gap, 0.0)\n    se = np.sqrt((vars_[:-1]**2 / np.array([len(r) for r in resids[:-1]]) +\n                  vars_[1:]**2 /\n                  np.array([len(r) for r in resids[1:]])) / 2.0) * 2.0\n    return float(np.max(ex / np.maximum(se, 1e-12)))\n\n\n# ------------------------------------------------------------- HSIC baseline\n\ndef hsic_stat(x, y, bw=None):\n    """Biased HSIC with RBF kernels (baseline CIT statistic)."""\n    x = np.asarray(x, float)\n    y = np.asarray(y, float)\n    n = len(x)\n    if bw is None:\n        sx = np.median(np.abs(x - np.median(x))) or 1.0\n        sy = np.median(np.abs(y - np.median(y))) or 1.0\n        bw_x, bw_y = max(sx * 1.06, 1e-6), max(sy * 1.06, 1e-6)\n    else:\n        bw_x, bw_y = bw\n    Kx = np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw_x) ** 2)\n    Ky = np.exp(-0.5 * ((y[:, None] - y[None, :]) / bw_y) ** 2)\n    Kxc = Kx - Kx.mean(axis=0) - Kx.mean(axis=1) + Kx.mean()\n    Kyc = Ky - Ky.mean(axis=0) - Ky.mean(axis=1) + Ky.mean()\n    return float((Kxc * Kyc).sum() / n ** 2)\n\n\ndef hsic_bootstrap(x, y, B=199, rng=None):\n    """Multiplier bootstrap for HSIC (Rademacher weights on feature\n    products\' low-rank approximation via direct reweighting).\n    DEPRECATED (v1): the quadratic-form draw xi\' P xi is not a valid null\n    distribution for the centered HSIC statistic (empirical size 1.0 under\n    all Phase-3 nulls); kept only for provenance. Use hsic_resid_stat +\n    hsic_resid_permutation instead."""\n    rng = rng or np.random.default_rng(20260829)\n    x = np.asarray(x, float)\n    y = np.asarray(y, float)\n    n = len(x)\n    sx = np.median(np.abs(x - np.median(x))) or 1.0\n    sy = np.median(np.abs(y - np.median(y))) or 1.0\n    Kx = np.exp(-0.5 * ((x[:, None] - x[None, :]) / max(sx * 1.06, 1e-6))**2)\n    Ky = np.exp(-0.5 * ((y[:, None] - y[None, :]) / max(sy * 1.06, 1e-6))**2)\n    Kxc = Kx - Kx.mean(axis=0) - Kx.mean(axis=1) + Kx.mean()\n    Kyc = Ky - Ky.mean(axis=0) - Ky.mean(axis=1) + Ky.mean()\n    prodc = Kxc * Kyc\n    draws = np.empty(B)\n    for b in range(B):\n        xi = rng.choice([-1.0, 1.0], size=n)\n        draws[b] = float(xi @ prodc @ xi) / n ** 2\n    return draws\n\n\n# ------------------------------------------------- v3 witness pipeline (D12)\n#\n# Post-hoc diagnosis of the Colab v2 calibration failure (heavy-tail sizes\n# ~1.0): (a) Rademacher sign-flips distort the marginal law of ASYMMETRIC\n# residuals (every draw is sampled from a pointwise-symmetrized law), so the\n# wild bootstrap targets the wrong functional; (b) constant-per-context\n# detrending leaves strata-width-induced shape artifacts that break any\n# label-exchangeability argument. v3 therefore:\n#   1. removes a per-context LINEAR trend (not a constant),\n#   2. winsorizes residuals within context (primary trim 0.05, predeclared\n#      grid {0, 0.01, 0.05} retained with 0.05 promoted to primary),\n#   3. affinely standardizes each context (center/scale by residual moments,\n#      as declared in the original docstring design),\n#   4. calibrates by CONTEXT-LABEL PERMUTATION of the pooled standardized\n#      residuals: an exact-multiset randomization null with no law\n#      distortion (valid under skewness/infinite moments), reusing the\n#      observed model\'s bandwidth (the pooled multiset is invariant).\n\n\ndef _ctx_linear_trend(x, y, ctx, Kr):\n    """Per-context linear trend fits; returns list of residual vectors."""\n    out = []\n    for c in range(Kr):\n        m = ctx == c\n        xs, ys = x[m], y[m]\n        if len(xs) >= 10 and np.std(xs) > 1e-12:\n            A = np.vstack([xs - xs.mean(), np.ones(len(xs))]).T\n            coef, *_ = np.linalg.lstsq(A, ys, rcond=None)\n            out.append(ys - A @ coef)\n        else:\n            out.append(ys - ys.mean())\n    return out\n\n\ndef _standardize_ctx(rs):\n    return [(r - r.mean()) / max(r.std(), 1e-9) for r in rs]\n\n\ndef _pooled_normal_score(zz):\n    """Global rank->Phi^{-1} transform of the pooled standardized\n    residuals (one fixed monotone bijection of the observed multiset).\n    Permuted-label draws inherit the exact same value multiset, so the\n    permutation null remains exact; the transform bounds the influence of\n    extreme order statistics on transport/kernel functionals."""\n    pool = np.concatenate(zz)\n    rnk = np.empty(len(pool))\n    rnk[np.argsort(pool, kind="stable")] = np.arange(len(pool))\n    from scipy.special import ndtri\n    u = (rnk + 0.5) / len(pool)\n    sc = ndtri(u)\n    out, ofs = [], 0\n    for z in zz:\n        out.append(sc[ofs:ofs + len(z)])\n        ofs += len(z)\n    return out\n\n\ndef _wz_std_resids(x, y, K=None, trim_q=0.05, detrend="linear",\n                   score=False):\n    """Full v3 residual preprocessing: trend removal, winsorization,\n    per-context standardization, optional pooled normal scoring.\n    Returns (list_of_vectors, lens, pool)."""\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    base = (_ctx_linear_trend(x, y, ctx, Kr) if detrend == "linear"\n            else context_residuals(x, y, ctx, Kr, trim_q=0.0))\n    rs = []\n    for r in base:\n        if trim_q > 0 and len(r) >= 10:\n            lo, hi = np.quantile(r, [trim_q, 1 - trim_q])\n            r = np.clip(r, lo, hi)\n        rs.append(r)\n    zz = _standardize_ctx(rs)\n    if score:\n        zz = _pooled_normal_score(zz)\n    return zz, [len(z) for z in zz], np.concatenate(zz)\n\n\ndef k1_v3_stat(x, y, trim_q=0.05, K=None, score=False):\n    """Point statistic T_K1 under the v3 pipeline (label-swap kernel QP)."""\n    zz, _, _ = _wz_std_resids(x, y, K=K, trim_q=trim_q, score=score)\n    return float(K1Witness(zz).solve())\n\n\ndef k2_v3_stat(x, y, trim_q=0.05, K=None, score=False):\n    """Point statistic T_K2 under the v3 pipeline (excess transport)."""\n    zz, _, _ = _wz_std_resids(x, y, K=K, trim_q=trim_q, score=score)\n    return k2_from_resids(zz)\n\n\ndef _perm_pool_draws(zz, lens, pool, bw, B, rng, smo_rounds=80):\n    draws = np.empty(B)\n    idx_split = np.cumsum(lens)[:-1]\n    for b in range(B):\n        parts = np.split(rng.permutation(pool), idx_split)\n        m = K1Witness(parts, bw=bw)\n        draws[b] = m.solve(smo_rounds=smo_rounds)\n    return draws\n\n\ndef k1_k2_perm_calibration(x, y, B=199, trims=(0.05,), K=None, rng=None,\n                           bmap=None, score=False):\n    """Joint label-permutation calibration for T_K1/T_K2.\n\n    Returns (obs_dict, draws_dict) where keys are trimming levels and\n    entries are dicts {\'k1\': stat, \'k2\': stat} / {\'k1\': arr, \'k2\': arr}.\n    bmap: optional {trim_q: B_eff} per-trim draw budgets."""\n    rng = rng or np.random.default_rng(20260901)\n    ctx = make_contexts(x, K=K)\n    Kr = len(np.unique(ctx))\n    out_obs, out_dr = {}, {}\n    for tq in trims:\n        zz, lens, pool = _wz_std_resids(x, y, K=K, trim_q=tq, score=score)\n        mo = K1Witness(zz)\n        bw = mo.bw                      # pooled law invariant under swap\n        o1 = float(mo.solve())\n        o2 = k2_from_resids(zz)\n        Beff = B if bmap is None else int(bmap.get(tq, B))\n        d1 = _perm_pool_draws(zz, lens, pool, bw, Beff, rng)\n        d2 = np.empty(Beff)\n        prng = np.random.default_rng(int(rng.integers(0, 2**62)))\n        idx_split = np.cumsum(lens)[:-1]\n        for b in range(Beff):\n            parts = np.split(prng.permutation(pool), idx_split)\n            d2[b] = k2_from_resids(parts)\n        out_obs[tq] = {"k1": o1, "k2": float(o2)}\n        out_dr[tq] = {"k1": d1, "k2": d2}\n    return out_obs, out_dr\n\n\ndef hsic_pairs_bootstrap(x, y, B=199, rng=None, cap=800):\n    """Pairs-bootstrap null draws for |residual-HSIC| (v3 baseline).\n\n    Resamples (x, y) pairs jointly and recomputes the whole NW-residual\n    HSIC pipeline; valid under H0 because the fitting artifact is carried\n    into every replicate. Returns abs-valued draws; pair the observed\n    value through abs(hsic_resid_stat(...))."""\n    rng = rng or np.random.default_rng(20260902)\n    x = np.asarray(x, float)[:cap]\n    y = np.asarray(y, float)[:cap]\n    n = len(x)\n    draws = np.empty(B)\n    for b in range(B):\n        ii = rng.integers(0, n, n)\n        xb, yb = x[ii], y[ii]\n        rb = _nw_residuals(xb, yb)\n        sx = np.median(np.abs(xb - np.median(xb))) or 1.0\n        sy = np.median(np.abs(rb - np.median(rb))) or 1.0\n        draws[b] = abs(hsic_stat(xb, rb,\n                                 bw=(max(sx * 1.06, 1e-6),\n                                     max(sy * 1.06, 1e-6))))\n    return draws\n\n\n# ---------------------------------------------- residual-CIT baseline (v2)\n\ndef _nw_residuals(x, y):\n    """Nadaraya-Watson residuals of y on x (Gaussian kernel, Silverman bw)."""\n    x = np.asarray(x, float)\n    y = np.asarray(y, float)\n    sx = np.median(np.abs(x - np.median(x))) or 1.0\n    bw = max(sx * 1.06 * len(x) ** (-0.2), 1e-6)\n    W = np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw) ** 2)\n    yhat = (W @ y) / np.maximum(W.sum(axis=1), 1e-12)\n    return y - yhat\n\n\ndef hsic_resid_stat(x, y, bw=None):\n    """Residual conditional-independence baseline (KCI-flavored): HSIC(X,\n    Y - g_hat(X)) with nonparametric g_hat. Tests noise exogeneity rather\n    than raw X-independence (the latter is false under both null and\n    alternative in the Phase-3 DGPs since X -> Y directly)."""\n    r = _nw_residuals(x, y)\n    return hsic_stat(x, r, bw=bw)\n\n\ndef hsic_resid_permutation(x, y, B=199, rng=None):\n    """Permutation null draws for hsic_resid_stat: permute the fitted\n    residuals (exchangeable under H0 given the mean fit). obs and draws\n    must be computed at the SAME sample size (driver caps both).\n    Kx is computed once and reused across permutations."""\n    rng = rng or np.random.default_rng(20260830)\n    x = np.asarray(x, float)\n    y = np.asarray(y, float)\n    r = _nw_residuals(x, y)\n    n = len(x)\n    sx = np.median(np.abs(x - np.median(x))) or 1.0\n    sr = np.median(np.abs(r - np.median(r))) or 1.0\n    bw_x = max(sx * 1.06, 1e-6)\n    bw_r = max(sr * 1.06, 1e-6)\n\n    def center(K):\n        return K - K.mean(axis=0) - K.mean(axis=1) + K.mean()\n\n    Kxc = center(np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw_x) ** 2))\n    draws = np.empty(B)\n    for b in range(B):\n        rp = rng.permutation(r)\n        Kyr = center(np.exp(-0.5 *\n                    ((rp[:, None] - rp[None, :]) / bw_r) ** 2))\n        draws[b] = float((Kxc * Kyr).sum()) / n ** 2\n    return draws\n\n"""Phase 3 seeded DGPs (WP 3.2 / 3.3).\n\nNulls (unconfounded within the declared additive-noise class):\n- null_gauss    : X = a*xi + e_X ; Y = c*X + e_Y (independent noises)\n- null_nonparam : Y = g(X) + e_Y with smooth seeded g\n\nAlternatives (B2 section 2.2 family - support of (X,Y) is R^2,\ncontractible, persistent-homology blind by construction):\n- conf_lin      : X = U + e_X ; Y = b*X + U + e_Y\n- conf_nonlin   : X = U + e_X ; Y = b*X + sin(1.5 U) + 0.5 U + e_Y\n\nNoise families for e_*: gauss, t3 (student-t df=3, standardized),\nlognorm (standardized). Nuisance covariates W extend dimension to d.\nAll generators return (X, Y, W) arrays with W possibly zero-width.\n"""\nimport numpy as np\n\n\ndef _noise(rng, n, kind):\n    if kind == "gauss":\n        return rng.normal(size=n)\n    if kind == "t3":\n        z = rng.standard_t(3, size=n)\n        return (z - z.mean()) / max(z.std(), 1e-9)\n    if kind == "lognorm":\n        z = rng.lognormal(size=n)\n        return (z - z.mean()) / max(z.std(), 1e-9)\n    raise ValueError(kind)\n\n\ndef _nuisance(rng, n, d):\n    return rng.normal(size=(n, max(d - 2, 0)))\n\n\ndef sample_null(rng, n, d, noise="gauss", kind="null_gauss"):\n    a = rng.choice([-2.0, -1.0, 1.0, 2.0])\n    c = rng.choice([-1.5, -1.0, 1.0, 1.5])\n    xi = rng.normal(size=n)\n    ex = _noise(rng, n, noise)\n    ey = _noise(rng, n, noise)\n    x = a * 0.7 * xi + ex\n    if kind == "null_gauss":\n        y = c * x + ey\n    elif kind == "null_nonparam":\n        y = 0.8 * np.tanh(x) + 0.4 * np.sin(x) + ey\n    else:\n        raise ValueError(kind)\n    W = _nuisance(rng, n, d)\n    return x, y, W\n\n\ndef sample_confounded(rng, n, d, b, noise="gauss", kind="conf_lin",\n                      u_scale=1.0):\n    u = rng.normal(size=n) * u_scale\n    ex = _noise(rng, n, noise)\n    ey = _noise(rng, n, noise)\n    x = u + ex\n    if kind == "conf_lin":\n        y = b * x + u + ey\n    elif kind == "conf_nonlin":\n        y = b * x + np.sin(1.5 * u) + 0.5 * u + ey\n    else:\n        raise ValueError(kind)\n    W = _nuisance(rng, n, d)\n    return x, y, W\n\n\ndef population_contextuality_probe(kind, b=0.0, noise="gauss", n=int(4e6),\n                                   seed=999, K=5):\n    """Large-sample witness value approximating the population functional\n    (used to classify contamination cells and check witness monotonicity\n    without sampling noise)."""\n    rng = np.random.default_rng(seed)\n    if kind.startswith("null"):\n        x, y, _ = sample_null(rng, n, 2, noise=noise, kind=kind)\n    else:\n        x, y, _ = sample_confounded(rng, n, 2, b, noise=noise, kind=kind)\n    return x, y\n\n\ndef critical_values(stat_draws, alpha_grid):\n    s = np.sort(np.asarray(stat_draws, dtype=float))\n    out = {}\n    for a in alpha_grid:\n        idx = min(max(int(np.ceil((1.0 - a) * len(s))) - 1, 0), len(s) - 1)\n        out[float(a)] = float(s[idx])\n    return out\n\n\nTRIMS = (0.0, 0.01, 0.05)\nALPHA_GRID = [round(0.01 * a, 2) for a in range(1, 21)]\nHSIC_CAP = 800\n# ---- WP 3.2 null-calibration driver (v3: linear detrend + context\n# standardization + label-permutation calibration; HSIC pairs-bootstrap).\nRESUME = os.path.exists(OUT)\ndone_seeds = set()\nif RESUME:\n    prev = pd.read_csv(OUT)\n    done_seeds = set((r["kind"], r["noise"], r["n"], r["d"], r["seed"])\n                     for _, r in prev.iterrows())\n    print("resume:", len(done_seeds), "dataset-records found")\n\nrows_all = []\nfor gi, g in enumerate(CONFIG["groups"]):\n    n, d, noise, kind, B = g["n"], g["d"], g["noise"], g["kind"], g["B"]\n    # v3 trimming policy: primary 0.05 at full B; sensitivity trims on\n    # gauss-noise nulls only, at reduced draw budgets.\n    if noise == "gauss":\n        trims = (0.0, 0.01, 0.05)\n        bmap = {0.0: min(B, 49), 0.01: min(B, 49), 0.05: B}\n    else:\n        trims = (0.05,)\n        bmap = None\n    todo = [s for s in g["seeds"]\n            if (kind, noise, n, d, s) not in done_seeds]\n    print("[group %d/%d] n=%d d=%d %s %s: %d seeds"\n          % (gi + 1, len(CONFIG["groups"]), n, d, noise, kind,\n             len(todo)), flush=True)\n    for j, seed in enumerate(todo):\n        rng = np.random.default_rng(seed)\n        x, y, _W = sample_null(rng, n, d, noise=noise, kind=kind)\n        obs, dr = k1_k2_perm_calibration(\n            x, y, B=B, trims=trims, bmap=bmap,\n            rng=np.random.default_rng(seed + 7000000))\n        xc, yc = x[:HSIC_CAP], y[:HSIC_CAP]\n        obs_hsic = abs(hsic_resid_stat(xc, yc))\n        hb = hsic_pairs_bootstrap(xc, yc, B=B,\n                                  rng=np.random.default_rng(\n                                      seed + 7200000), cap=HSIC_CAP)\n        for meth in ("k1", "k2"):\n            for tq in trims:\n                cvs = critical_values(dr[tq][meth], ALPHA_GRID)\n                r = {"n": n, "d": d, "noise": noise, "kind": kind,\n                     "seed": seed, "method": meth, "trim": tq,\n                     "B": len(dr[tq][meth]),\n                     "stat_obs": obs[tq][meth]}\n                for a in ALPHA_GRID:\n                    r["cv_%.2f" % a] = cvs[a]\n                rows_all.append(r)\n        cvs = critical_values(hb, ALPHA_GRID)\n        r = {"n": n, "d": d, "noise": noise, "kind": kind,\n             "seed": seed, "method": "hsic", "trim": 0.0,\n             "B": len(hb), "stat_obs": obs_hsic}\n        for a in ALPHA_GRID:\n            r["cv_%.2f" % a] = cvs[a]\n        rows_all.append(r)\n        if (j + 1) % 10 == 0 or (j + 1) == len(todo):\n            pd.DataFrame(rows_all).to_csv(OUT, index=False)\n            print("  %d/%d seeds, rows=%d" % (j + 1, len(todo),\n                                              len(rows_all)), flush=True)\npd.DataFrame(rows_all).to_csv(OUT, index=False)\n\nmanifest = {"tag": TAG, "shard_id": SHARD_ID, "code_hash": CODE_HASH,\n            "groups": len(CONFIG["groups"]),\n            "rows_written": len(rows_all)}\nmpath = os.path.join(os.path.dirname(OUT),\n                     "ccx_%s_manifest_shard%02d.json" % (TAG, SHARD_ID))\nwith open(mpath, "w") as fh:\n    json.dump(manifest, fh, indent=2)\nprint("MANIFEST:", json.dumps(manifest))\n\ntry:\n    from google.colab import files\n    files.download(OUT)\n    files.download(mpath)\n    print("Downloaded:", OUT)\nexcept Exception as e:\n    print("(Not on Colab / download skipped):", e)\n').encode()).hexdigest()[:16]
print(TAG, 'shard', SHARD_ID)
import os, sys, time, json, hashlib
import numpy as np
import pandas as pd

"""WP 3.1 continuous-outcome witnesses (Phase 3 flagship, claim C4).

Constructions (predeclared design; see memos/gate_C_memo.md):

Context cover: X-strata (quantile bins of the treatment/residual covariate),
K strata by n (K = 3..8). Within each context c, ADDITIVE-MECHANISM
residuals r = y - m_hat(x) form an empirical law mu_c.

Declared structural null S_cont (homoskedastic shared-shape ANM):
y = m(x) + sigma * eps, eps ~ F shared, eps independent of x. After
per-context affine standardization (center/scale by context residual
moments), every context law must COINCIDE with one common shape. Latent
common causes break this: U enters both arms, so stratum-restricted U
ranges inject stratum-dependent variance/skew into standardized residuals.

K1 (kernelized forcing-divergence witness):
  T_K1 = min_{w in simplex} sum_c MMD_k^2( mu_hat_c , nu_c(w) ),
where nu_c(w) is the Gaussian-smoothed mixture with weights w over a fixed
component grid t_1..t_G (standard-normal quantiles) pushed through the
context affine map. Mean embeddings are linear in w => exact QP per sum;
solved by projected gradient (Frank-Wolfe) to machine tolerance. Value 0
iff ONE shared coupling explains all context laws simultaneously (in the
kernel-smoothed sense) - the continuous descendant of the discrete KL
contextuality functional (KL -> MMD^2 forcing divergence).

K2 (transport-defect witness):
  Consecutive-context squared Wasserstein distances d_c =
  W2^2(mu_c, mu_{c+1}) (exact 1-D empirical OT). Under the shared-shape
  null with drift, d_c ~ (m_{c+1} - m_c)^2 up to sampling error; the
  witness is the max standardized EXCESS transport
  T_K2 = max_c (d_c - fitted)/se, the cycle-space/monodromy-excess
  descendant of B2's topological statistic, requiring no support holes.

Calibration: multiplier (Rademacher wild) bootstrap recomputing the whole
pipeline on multiplier-perturbed residuals, including the inner
optimization; optional trimming policy applied BEFORE context moments
(predeclared grids q in {0, 0.01, 0.05}).

All estimators take arrays (x, y) and return scalar statistics; bootstrap
variants return draws for critical values.
"""
import numpy as np

try:
    import ot as pot
except Exception:                                   # pragma: no cover
    pot = None


# ------------------------------------------------------------------ contexts

def make_contexts(x, K=None, return_edges=False):
    """Quantile-stratify x into K contexts. Default K by n: 3 (<800),
    5 (<4000), 8 otherwise (predeclared)."""
    x = np.asarray(x, dtype=float)
    n = len(x)
    if K is None:
        K = 3 if n < 800 else (5 if n < 4000 else 8)
    qs = np.linspace(0, 1, K + 1)[1:-1]
    edges = np.quantile(x, qs)
    ctx = np.searchsorted(edges, x, side="right")
    if return_edges:
        return ctx, edges, K
    return ctx


def context_residuals(x, y, ctx, K, trim_q=0.0):
    """Additive-mechanism residuals per context: r = y - context-mean of y
    (context-constant m_hat; documented estimator). Winsorizes residuals at
    trim_q/1-trim_q WITHIN context when trim_q > 0."""
    out = []
    for c in range(K):
        m = ctx == c
        r = y[m] - y[m].mean()
        if trim_q > 0 and len(r) >= 10:
            lo, hi = np.quantile(r, [trim_q, 1 - trim_q])
            r = np.clip(r, lo, hi)
        out.append(r)
    return out


def _median_bandwidth(resids):
    pool = np.concatenate(resids)
    pool = pool[np.isfinite(pool)]
    if len(pool) < 10:
        return 1.0
    sub = pool[::max(1, len(pool) // 2000)]
    d = np.abs(sub[:, None] - sub[None, :])
    med = np.median(d[d > 0]) if np.any(d > 0) else 1.0
    return max(med * 1.06, 1e-6)


def _gram(points, bw):
    """RBF Gram matrix and cross matrices against component grid."""
    z = (points[:, None] - points[None, :]) / bw
    return np.exp(-0.5 * z ** 2)


# ------------------------------------------------------------------ K1 core

_GRID_CACHE = {}


def _component_grid(G):
    """Standard-normal quantile grid (deterministic rng(7)); cached per G
    since it is identical across every construction."""
    if G not in _GRID_CACHE:
        _GRID_CACHE[G] = np.quantile(
            np.random.default_rng(7).normal(size=200000),
            (np.arange(G) + 0.5) / G)
    return _GRID_CACHE[G]


class K1Witness:
    """Kernelized forcing-divergence witness over a shared coupling grid."""

    def __init__(self, resids, G=32, bw=None, fw_iters=300, m_cap=600):
        self.K = len(resids)
        self.G = G
        self.fw_iters = fw_iters
        self.bw = bw if bw is not None else _median_bandwidth(resids)
        self.m_cap = m_cap
        # component grid: standard normal quantiles
        self.t = _component_grid(G)
        # per-context mean embeddings against grid components and self-Gram
        self.b = []      # b_c[j] = <mu_c, nu_j>
        self.A = []      # A_c[j,j'] = <nu_j, nu_j'> (same for all c)
        self.self_mmd_const = []
        Ky = _gram(self.t, self.bw)
        self.A = Ky.copy()
        for r in resids:
            if len(r) == 0:
                r = np.zeros(1)
            rs = r[:min(len(r), self.m_cap)]
            k_cross = np.exp(-0.5 * ((rs[:, None] - self.t[None, :]) /
                                     self.bw) ** 2).mean(axis=0)
            self.b.append(k_cross)
            Krr = _gram(rs[:min(len(rs), 400)], self.bw)
            self.self_mmd_const.append(float(Krr.mean()))

    def objective(self, w, c=None):
        """sum_c MMD^2(mu_c, nu_c(w)) (constants included)."""
        idx = range(self.K) if c is None else [c]
        tot = 0.0
        for cc in idx:
            tot += float(w @ self.A @ w - 2 * self.b[cc] @ w +
                         self.self_mmd_const[cc])
        return tot

    def solve(self, w_init=None, smo_rounds=None):
        """Frank-Wolfe on the simplex for min_w sum_c MMD^2(mu_c, nu_c(w)).
        Objective (up to constants): f(w) = K * w'Aw - 2 * (sum_c b_c)'w.
        Exact line-search FW followed by pairwise (SMO) refinement."""
        s = np.sum(self.b, axis=0)
        w = np.ones(self.G) / self.G if w_init is None else \
            np.asarray(w_init, float).copy()
        w = np.maximum(w, 0)
        w /= max(w.sum(), 1e-300)
        prev = np.inf
        for k in range(self.fw_iters):
            grad = 2 * self.K * (self.A @ w) - 2 * s
            j = int(np.argmin(grad))
            d = np.zeros(self.G)
            d[j] = 1.0
            # exact step via line search on the segment (convex quadratic):
            num = float(grad @ (d - w))
            den = float(2 * self.K * (d - w) @ self.A @ (d - w))
            gamma = 0.0 if den <= 0 else max(0.0, min(1.0,
                                                      -num / den))
            if gamma == 0.0:
                break
            w = (1 - gamma) * w + gamma * d
            val = self.objective(w)
            if abs(prev - val) < 1e-14 * max(1.0, abs(val)):
                break
            prev = val
        w = self._pairwise_polish(w, max_rounds=smo_rounds or 300)
        self.w_star = np.maximum(w, 0)
        self.w_star /= max(self.w_star.sum(), 1e-300)
        self.value = max(self.objective(self.w_star), 0.0)
        return self.value

    def _pairwise_polish(self, w, max_rounds=300, tol=1e-15):
        """Exact pairwise coordinate optimization over the simplex:
        repeatedly move mass along the best pair direction e_i - e_j with
        closed-form step, guaranteeing monotone descent to the QP optimum."""
        s = np.sum(self.b, axis=0)
        K = self.K
        prev = self.objective(w)
        for _ in range(max_rounds):
            g = 2 * K * (self.A @ w - s / K)
            # d_ij = e_i - e_j ; decrease along gamma:
            # f(w + g*d) = f + gamma*(g_i - g_j) + gamma^2 * K * c_ij
            Gd = self.A  # quadratic coefficients via diag/offdiag
            gi = g[:, None] - g[None, :]
            cij = np.diag(Gd)[:, None] + np.diag(Gd)[None, :] \
                - 2 * Gd
            den = 2 * K * np.maximum(cij, 0.0)
            with np.errstate(divide="ignore", invalid="ignore"):
                gam = -gi / np.where(den > 0, den, 1.0)
            hi = np.tile(w, (self.G, 1))          # gamma <= w_j
            lo = -np.tile(w[:, None], (1, self.G))  # gamma >= -w_i
            gam = np.clip(np.where(den > 0, gam, 0.0),
                          np.minimum(lo, 0.0) - 1e-12,
                          np.maximum(hi, 0.0) + 1e-12)
            dec = gi * gam + 0.5 * den * gam ** 2
            np.fill_diagonal(dec, 0.0)
            i, j = np.unravel_index(np.argmin(dec), dec.shape)
            best_dec = -dec[i, j]
            if best_dec <= tol * max(1.0, abs(prev)):
                break
            gamma = gam[i, j]
            w = w.copy()
            w[i] += gamma
            w[j] -= gamma
            prev -= best_dec
        return w


def k1_witness(x, y, trim_q=0.0, G=32, K=None, return_model=False):
    """Point statistic T_K1."""
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    resids = context_residuals(x, y, ctx, Kr, trim_q=trim_q)
    model = K1Witness(resids, G=G)
    val = model.solve()
    if return_model:
        return val, model
    return val


def k1_multiplier_bootstrap(x, y, B=199, trim_grid=(0.0,), G_boot=16, K=None,
                            rng=None, bmap=None):
    """Wild-bootstrap draws of T_K1 (Rademacher multipliers on residuals,
    pipeline re-run incl. inner optimization). bmap: optional
    {trim_q: B_eff} per-trim draw budgets (D8). Returns (B,) array for the
    FIRST trimming level plus dict of arrays per trimming level."""
    rng = rng or np.random.default_rng(20260827)
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    base_resids = context_residuals(x, y, ctx, Kr, trim_q=0.0)
    base_model = K1Witness(base_resids, G=G_boot)
    base_model.solve()
    w_star = base_model.w_star
    draws = {}
    for tq in trim_grid:
        b_eff = B if bmap is None else int(bmap.get(tq, B))
        vals = np.empty(b_eff)
        for b in range(b_eff):
            pert = [(r * rng.choice([-1.0, 1.0], size=len(r))) for r in
                    base_resids]
            if tq > 0:
                pert = [_winsor(r, tq) for r in pert]
            m = K1Witness(pert, G=G_boot, bw=base_model.bw)
            vals[b] = m.solve(w_init=w_star, smo_rounds=80)
        draws[tq] = vals
    return draws[list(trim_grid)[0]], draws


def _winsor(r, q):
    lo, hi = np.quantile(r, [q, 1 - q])
    return np.clip(r, lo, hi)


# ------------------------------------------------------------------ K2 core

def _w2sq(a, b):
    """Exact 1-D empirical squared Wasserstein-2 (sort-based)."""
    a = np.sort(np.asarray(a, float))
    b = np.sort(np.asarray(b, float))
    n = min(len(a), len(b))
    if n < 2:
        return 0.0
    ai = np.quantile(a, np.linspace(0, 1, n))
    bi = np.quantile(b, np.linspace(0, 1, n))
    return float(((ai - bi) ** 2).mean())


def k2_witness(x, y, trim_q=0.0, K=None, return_parts=False):
    """Transport-defect witness: max standardized excess of consecutive-
    context W2^2 over the location-gap prediction."""
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    resids = context_residuals(x, y, ctx, Kr, trim_q=trim_q)
    means = np.array([r.mean() for r in resids])
    vars_ = np.array([r.var() for r in resids])
    d = np.array([_w2sq(resids[c], resids[c + 1]) for c in range(Kr - 1)])
    gap = np.diff(means) ** 2
    if len(d) < 2:
        val = 0.0
    else:
        # excess over drift prediction, standardized by pooled sampling scale
        resid_excess = np.maximum(d - gap, 0.0)
        se = np.sqrt((vars_[:-1]**2 / np.array([len(r) for r in resids[:-1]])
                      + vars_[1:]**2 /
                      np.array([len(r) for r in resids[1:]])) / 2.0) * 2.0
        se = np.maximum(se, 1e-12)
        val = float(np.max(resid_excess / se))
    if return_parts:
        return val, {"d": d, "gap": gap, "means": means, "vars": vars_}
    return val


def k2_multiplier_bootstrap(x, y, B=199, trim_grid=(0.0,), K=None,
                            rng=None, bmap=None):
    """bmap: optional {trim_q: B_eff} per-trim draw budgets (D8)."""
    rng = rng or np.random.default_rng(20260828)
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    base = context_residuals(x, y, ctx, Kr, trim_q=0.0)
    draws = {}
    for tq in trim_grid:
        b_eff = B if bmap is None else int(bmap.get(tq, B))
        vals = np.empty(b_eff)
        for b in range(b_eff):
            pert = [(r * rng.choice([-1.0, 1.0], size=len(r))) for r in base]
            if tq > 0:
                pert = [_winsor(r, tq) for r in pert]
            vals[b] = k2_from_resids(pert)
        draws[tq] = vals
    return draws[list(trim_grid)[0]], draws


def k2_from_resids(resids):
    means = np.array([r.mean() for r in resids])
    vars_ = np.array([r.var() for r in resids])
    d = np.array([_w2sq(resids[c], resids[c + 1])
                  for c in range(len(resids) - 1)])
    gap = np.diff(means) ** 2
    if len(d) < 2:
        return 0.0
    ex = np.maximum(d - gap, 0.0)
    se = np.sqrt((vars_[:-1]**2 / np.array([len(r) for r in resids[:-1]]) +
                  vars_[1:]**2 /
                  np.array([len(r) for r in resids[1:]])) / 2.0) * 2.0
    return float(np.max(ex / np.maximum(se, 1e-12)))


# ------------------------------------------------------------- HSIC baseline

def hsic_stat(x, y, bw=None):
    """Biased HSIC with RBF kernels (baseline CIT statistic)."""
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    n = len(x)
    if bw is None:
        sx = np.median(np.abs(x - np.median(x))) or 1.0
        sy = np.median(np.abs(y - np.median(y))) or 1.0
        bw_x, bw_y = max(sx * 1.06, 1e-6), max(sy * 1.06, 1e-6)
    else:
        bw_x, bw_y = bw
    Kx = np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw_x) ** 2)
    Ky = np.exp(-0.5 * ((y[:, None] - y[None, :]) / bw_y) ** 2)
    Kxc = Kx - Kx.mean(axis=0) - Kx.mean(axis=1) + Kx.mean()
    Kyc = Ky - Ky.mean(axis=0) - Ky.mean(axis=1) + Ky.mean()
    return float((Kxc * Kyc).sum() / n ** 2)


def hsic_bootstrap(x, y, B=199, rng=None):
    """Multiplier bootstrap for HSIC (Rademacher weights on feature
    products' low-rank approximation via direct reweighting).
    DEPRECATED (v1): the quadratic-form draw xi' P xi is not a valid null
    distribution for the centered HSIC statistic (empirical size 1.0 under
    all Phase-3 nulls); kept only for provenance. Use hsic_resid_stat +
    hsic_resid_permutation instead."""
    rng = rng or np.random.default_rng(20260829)
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    n = len(x)
    sx = np.median(np.abs(x - np.median(x))) or 1.0
    sy = np.median(np.abs(y - np.median(y))) or 1.0
    Kx = np.exp(-0.5 * ((x[:, None] - x[None, :]) / max(sx * 1.06, 1e-6))**2)
    Ky = np.exp(-0.5 * ((y[:, None] - y[None, :]) / max(sy * 1.06, 1e-6))**2)
    Kxc = Kx - Kx.mean(axis=0) - Kx.mean(axis=1) + Kx.mean()
    Kyc = Ky - Ky.mean(axis=0) - Ky.mean(axis=1) + Ky.mean()
    prodc = Kxc * Kyc
    draws = np.empty(B)
    for b in range(B):
        xi = rng.choice([-1.0, 1.0], size=n)
        draws[b] = float(xi @ prodc @ xi) / n ** 2
    return draws


# ------------------------------------------------- v3 witness pipeline (D12)
#
# Post-hoc diagnosis of the Colab v2 calibration failure (heavy-tail sizes
# ~1.0): (a) Rademacher sign-flips distort the marginal law of ASYMMETRIC
# residuals (every draw is sampled from a pointwise-symmetrized law), so the
# wild bootstrap targets the wrong functional; (b) constant-per-context
# detrending leaves strata-width-induced shape artifacts that break any
# label-exchangeability argument. v3 therefore:
#   1. removes a per-context LINEAR trend (not a constant),
#   2. winsorizes residuals within context (primary trim 0.05, predeclared
#      grid {0, 0.01, 0.05} retained with 0.05 promoted to primary),
#   3. affinely standardizes each context (center/scale by residual moments,
#      as declared in the original docstring design),
#   4. calibrates by CONTEXT-LABEL PERMUTATION of the pooled standardized
#      residuals: an exact-multiset randomization null with no law
#      distortion (valid under skewness/infinite moments), reusing the
#      observed model's bandwidth (the pooled multiset is invariant).


def _ctx_linear_trend(x, y, ctx, Kr):
    """Per-context linear trend fits; returns list of residual vectors."""
    out = []
    for c in range(Kr):
        m = ctx == c
        xs, ys = x[m], y[m]
        if len(xs) >= 10 and np.std(xs) > 1e-12:
            A = np.vstack([xs - xs.mean(), np.ones(len(xs))]).T
            coef, *_ = np.linalg.lstsq(A, ys, rcond=None)
            out.append(ys - A @ coef)
        else:
            out.append(ys - ys.mean())
    return out


def _standardize_ctx(rs):
    return [(r - r.mean()) / max(r.std(), 1e-9) for r in rs]


def _pooled_normal_score(zz):
    """Global rank->Phi^{-1} transform of the pooled standardized
    residuals (one fixed monotone bijection of the observed multiset).
    Permuted-label draws inherit the exact same value multiset, so the
    permutation null remains exact; the transform bounds the influence of
    extreme order statistics on transport/kernel functionals."""
    pool = np.concatenate(zz)
    rnk = np.empty(len(pool))
    rnk[np.argsort(pool, kind="stable")] = np.arange(len(pool))
    from scipy.special import ndtri
    u = (rnk + 0.5) / len(pool)
    sc = ndtri(u)
    out, ofs = [], 0
    for z in zz:
        out.append(sc[ofs:ofs + len(z)])
        ofs += len(z)
    return out


def _wz_std_resids(x, y, K=None, trim_q=0.05, detrend="linear",
                   score=False):
    """Full v3 residual preprocessing: trend removal, winsorization,
    per-context standardization, optional pooled normal scoring.
    Returns (list_of_vectors, lens, pool)."""
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    base = (_ctx_linear_trend(x, y, ctx, Kr) if detrend == "linear"
            else context_residuals(x, y, ctx, Kr, trim_q=0.0))
    rs = []
    for r in base:
        if trim_q > 0 and len(r) >= 10:
            lo, hi = np.quantile(r, [trim_q, 1 - trim_q])
            r = np.clip(r, lo, hi)
        rs.append(r)
    zz = _standardize_ctx(rs)
    if score:
        zz = _pooled_normal_score(zz)
    return zz, [len(z) for z in zz], np.concatenate(zz)


def k1_v3_stat(x, y, trim_q=0.05, K=None, score=False):
    """Point statistic T_K1 under the v3 pipeline (label-swap kernel QP)."""
    zz, _, _ = _wz_std_resids(x, y, K=K, trim_q=trim_q, score=score)
    return float(K1Witness(zz).solve())


def k2_v3_stat(x, y, trim_q=0.05, K=None, score=False):
    """Point statistic T_K2 under the v3 pipeline (excess transport)."""
    zz, _, _ = _wz_std_resids(x, y, K=K, trim_q=trim_q, score=score)
    return k2_from_resids(zz)


def _perm_pool_draws(zz, lens, pool, bw, B, rng, smo_rounds=80):
    draws = np.empty(B)
    idx_split = np.cumsum(lens)[:-1]
    for b in range(B):
        parts = np.split(rng.permutation(pool), idx_split)
        m = K1Witness(parts, bw=bw)
        draws[b] = m.solve(smo_rounds=smo_rounds)
    return draws


def k1_k2_perm_calibration(x, y, B=199, trims=(0.05,), K=None, rng=None,
                           bmap=None, score=False):
    """Joint label-permutation calibration for T_K1/T_K2.

    Returns (obs_dict, draws_dict) where keys are trimming levels and
    entries are dicts {'k1': stat, 'k2': stat} / {'k1': arr, 'k2': arr}.
    bmap: optional {trim_q: B_eff} per-trim draw budgets."""
    rng = rng or np.random.default_rng(20260901)
    ctx = make_contexts(x, K=K)
    Kr = len(np.unique(ctx))
    out_obs, out_dr = {}, {}
    for tq in trims:
        zz, lens, pool = _wz_std_resids(x, y, K=K, trim_q=tq, score=score)
        mo = K1Witness(zz)
        bw = mo.bw                      # pooled law invariant under swap
        o1 = float(mo.solve())
        o2 = k2_from_resids(zz)
        Beff = B if bmap is None else int(bmap.get(tq, B))
        d1 = _perm_pool_draws(zz, lens, pool, bw, Beff, rng)
        d2 = np.empty(Beff)
        prng = np.random.default_rng(int(rng.integers(0, 2**62)))
        idx_split = np.cumsum(lens)[:-1]
        for b in range(Beff):
            parts = np.split(prng.permutation(pool), idx_split)
            d2[b] = k2_from_resids(parts)
        out_obs[tq] = {"k1": o1, "k2": float(o2)}
        out_dr[tq] = {"k1": d1, "k2": d2}
    return out_obs, out_dr


def hsic_pairs_bootstrap(x, y, B=199, rng=None, cap=800):
    """Pairs-bootstrap null draws for |residual-HSIC| (v3 baseline).

    Resamples (x, y) pairs jointly and recomputes the whole NW-residual
    HSIC pipeline; valid under H0 because the fitting artifact is carried
    into every replicate. Returns abs-valued draws; pair the observed
    value through abs(hsic_resid_stat(...))."""
    rng = rng or np.random.default_rng(20260902)
    x = np.asarray(x, float)[:cap]
    y = np.asarray(y, float)[:cap]
    n = len(x)
    draws = np.empty(B)
    for b in range(B):
        ii = rng.integers(0, n, n)
        xb, yb = x[ii], y[ii]
        rb = _nw_residuals(xb, yb)
        sx = np.median(np.abs(xb - np.median(xb))) or 1.0
        sy = np.median(np.abs(rb - np.median(rb))) or 1.0
        draws[b] = abs(hsic_stat(xb, rb,
                                 bw=(max(sx * 1.06, 1e-6),
                                     max(sy * 1.06, 1e-6))))
    return draws


# ---------------------------------------------- residual-CIT baseline (v2)

def _nw_residuals(x, y):
    """Nadaraya-Watson residuals of y on x (Gaussian kernel, Silverman bw)."""
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    sx = np.median(np.abs(x - np.median(x))) or 1.0
    bw = max(sx * 1.06 * len(x) ** (-0.2), 1e-6)
    W = np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw) ** 2)
    yhat = (W @ y) / np.maximum(W.sum(axis=1), 1e-12)
    return y - yhat


def hsic_resid_stat(x, y, bw=None):
    """Residual conditional-independence baseline (KCI-flavored): HSIC(X,
    Y - g_hat(X)) with nonparametric g_hat. Tests noise exogeneity rather
    than raw X-independence (the latter is false under both null and
    alternative in the Phase-3 DGPs since X -> Y directly)."""
    r = _nw_residuals(x, y)
    return hsic_stat(x, r, bw=bw)


def hsic_resid_permutation(x, y, B=199, rng=None):
    """Permutation null draws for hsic_resid_stat: permute the fitted
    residuals (exchangeable under H0 given the mean fit). obs and draws
    must be computed at the SAME sample size (driver caps both).
    Kx is computed once and reused across permutations."""
    rng = rng or np.random.default_rng(20260830)
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    r = _nw_residuals(x, y)
    n = len(x)
    sx = np.median(np.abs(x - np.median(x))) or 1.0
    sr = np.median(np.abs(r - np.median(r))) or 1.0
    bw_x = max(sx * 1.06, 1e-6)
    bw_r = max(sr * 1.06, 1e-6)

    def center(K):
        return K - K.mean(axis=0) - K.mean(axis=1) + K.mean()

    Kxc = center(np.exp(-0.5 * ((x[:, None] - x[None, :]) / bw_x) ** 2))
    draws = np.empty(B)
    for b in range(B):
        rp = rng.permutation(r)
        Kyr = center(np.exp(-0.5 *
                    ((rp[:, None] - rp[None, :]) / bw_r) ** 2))
        draws[b] = float((Kxc * Kyr).sum()) / n ** 2
    return draws

"""Phase 3 seeded DGPs (WP 3.2 / 3.3).

Nulls (unconfounded within the declared additive-noise class):
- null_gauss    : X = a*xi + e_X ; Y = c*X + e_Y (independent noises)
- null_nonparam : Y = g(X) + e_Y with smooth seeded g

Alternatives (B2 section 2.2 family - support of (X,Y) is R^2,
contractible, persistent-homology blind by construction):
- conf_lin      : X = U + e_X ; Y = b*X + U + e_Y
- conf_nonlin   : X = U + e_X ; Y = b*X + sin(1.5 U) + 0.5 U + e_Y

Noise families for e_*: gauss, t3 (student-t df=3, standardized),
lognorm (standardized). Nuisance covariates W extend dimension to d.
All generators return (X, Y, W) arrays with W possibly zero-width.
"""
import numpy as np


def _noise(rng, n, kind):
    if kind == "gauss":
        return rng.normal(size=n)
    if kind == "t3":
        z = rng.standard_t(3, size=n)
        return (z - z.mean()) / max(z.std(), 1e-9)
    if kind == "lognorm":
        z = rng.lognormal(size=n)
        return (z - z.mean()) / max(z.std(), 1e-9)
    raise ValueError(kind)


def _nuisance(rng, n, d):
    return rng.normal(size=(n, max(d - 2, 0)))


def sample_null(rng, n, d, noise="gauss", kind="null_gauss"):
    a = rng.choice([-2.0, -1.0, 1.0, 2.0])
    c = rng.choice([-1.5, -1.0, 1.0, 1.5])
    xi = rng.normal(size=n)
    ex = _noise(rng, n, noise)
    ey = _noise(rng, n, noise)
    x = a * 0.7 * xi + ex
    if kind == "null_gauss":
        y = c * x + ey
    elif kind == "null_nonparam":
        y = 0.8 * np.tanh(x) + 0.4 * np.sin(x) + ey
    else:
        raise ValueError(kind)
    W = _nuisance(rng, n, d)
    return x, y, W


def sample_confounded(rng, n, d, b, noise="gauss", kind="conf_lin",
                      u_scale=1.0):
    u = rng.normal(size=n) * u_scale
    ex = _noise(rng, n, noise)
    ey = _noise(rng, n, noise)
    x = u + ex
    if kind == "conf_lin":
        y = b * x + u + ey
    elif kind == "conf_nonlin":
        y = b * x + np.sin(1.5 * u) + 0.5 * u + ey
    else:
        raise ValueError(kind)
    W = _nuisance(rng, n, d)
    return x, y, W


def population_contextuality_probe(kind, b=0.0, noise="gauss", n=int(4e6),
                                   seed=999, K=5):
    """Large-sample witness value approximating the population functional
    (used to classify contamination cells and check witness monotonicity
    without sampling noise)."""
    rng = np.random.default_rng(seed)
    if kind.startswith("null"):
        x, y, _ = sample_null(rng, n, 2, noise=noise, kind=kind)
    else:
        x, y, _ = sample_confounded(rng, n, 2, b, noise=noise, kind=kind)
    return x, y


def critical_values(stat_draws, alpha_grid):
    s = np.sort(np.asarray(stat_draws, dtype=float))
    out = {}
    for a in alpha_grid:
        idx = min(max(int(np.ceil((1.0 - a) * len(s))) - 1, 0), len(s) - 1)
        out[float(a)] = float(s[idx])
    return out


TRIMS = (0.0, 0.01, 0.05)
ALPHA_GRID = [round(0.01 * a, 2) for a in range(1, 21)]
HSIC_CAP = 800
```
